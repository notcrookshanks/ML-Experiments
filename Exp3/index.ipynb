# Experiment 3: Categorical Naive Bayes Classification

**Objective:**
1. Create a toy dataset with categorical features (`Weather`, `Temperature`) and target (`Play`).
2. Encode categorical variables using `OrdinalEncoder` (features) and `LabelEncoder` (target).
3. Train a `CategoricalNB` classifier.
4. Evaluate accuracy and generate a classification report.
5. Predict the class for a new unseen scenario.

### Step 1: Import Libraries

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OrdinalEncoder, LabelEncoder
from sklearn.naive_bayes import CategoricalNB
from sklearn.metrics import accuracy_score, classification_report
```

### Step 2: Create Categorical Dataset

```python
# Create toy tennis/play dataset
df = pd.DataFrame({
    'Weather': ['Sunny', 'Rainy', 'Cloudy', 'Sunny', 'Rainy',
                'Cloudy', 'Sunny', 'Rainy', 'Cloudy', 'Sunny'],
    'Temperature': ['Hot', 'Cool', 'Mild', 'Mild', 'Hot',
                    'Cool', 'Cool', 'Mild', 'Hot', 'Hot'],
    'Play': ['No', 'Yes', 'Yes', 'No', 'Yes',
             'Yes', 'Yes', 'Yes', 'Yes', 'No']
})

print("Dataset:")
print(df)
```

### Step 3: Encode Categorical Features and Target

```python
# Separate features X and target y
X = df[['Weather', 'Temperature']]
y = df['Play']

# OrdinalEncoder converts feature strings to integers
feature_encoder = OrdinalEncoder()
X_encoded = feature_encoder.fit_transform(X)

# LabelEncoder converts target strings to 0/1
target_encoder = LabelEncoder()
y_encoded = target_encoder.fit_transform(y)

print("Encoded Features (X):\n", X_encoded[:5])
print("\nEncoded Target (y):\n", y_encoded[:5])
```

### Step 4: Split Data and Train Categorical Naive Bayes

```python
# 70% Train, 30% Test split
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y_encoded, test_size=0.3, random_state=42
)

# Initialize and train CategoricalNB
model = CategoricalNB()
model.fit(X_train, y_train)

print("Categorical Naive Bayes trained successfully.")
```

### Step 5: Evaluate the Classifier

```python
# Predict on test set
y_pred = model.predict(X_test)

# Accuracy & Report
acc = accuracy_score(y_test, y_pred)
print(f"Accuracy: {acc * 100:.2f}%\n")
print("Classification Report:")
print(classification_report(y_test, y_pred, zero_division=0))
```

### Step 6: Make Prediction on a New Scenario

```python
# Test sample: Weather = 'Sunny', Temperature = 'Cool'
new_sample = pd.DataFrame({
    'Weather': ['Sunny'],
    'Temperature': ['Cool']
})

# Encode sample using feature_encoder
new_sample_encoded = feature_encoder.transform(new_sample)

# Predict numeric class
pred_numeric = model.predict(new_sample_encoded)

# Decode back to label ('No' / 'Yes')
pred_label = target_encoder.inverse_transform(pred_numeric)

print("Scenario: Weather='Sunny', Temperature='Cool'")
print("Predicted Play Decision:", pred_label[0])
```
